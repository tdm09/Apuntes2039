### Handling the missing values

####  Método 1:

Descartar las observaciones con valores faltantes

```python
ndf = df.copy()
```

    ndf.dropna()   -  This method is used to remove missing values from a DataFrame.

  By default, it drops any row that contains $at\ least\ one$ missing value.

```python
ndf.dropna(inplace = True)
```

```python
ndf.isna().values.any()
```

```python
ndf
```

El problema con esta estrategia es que:
- si falta algún dato en todo el conjunto de datos, la fila correspondiente se elimina.

```python
ndf.isna().any()
```

####  Método 2:

Suelte las `columnas` donde falta `al menos` un elemento:

```python
ndf = df.copy()
ndf
```

Suelte las `columnas` donde falta al menos un elemento:

```python
ndf.dropna(axis = 1,      # axis 1 is acros the rows
           inplace = True)
ndf
```

####  Método 3:

Suelte las `columnas` donde falta `al menos` un elemento:

```python
ndf = df.copy()
ndf
```

Drop the rows where `all elements` are missing.

This method removes rows (or columns, depending on the axis parameter) from the DataFrame ndf where all entries are missing (NaN).

The how='all' parameter specifies that only rows where every value is NaN will be dropped.

```python
ndf.dropna(how='all',
           inplace = True)
ndf
```

### Método 4:

    Threshold  - thresh = 2  : `Mantenga` solo las filas con al menos 2 valores que  NO SEAN `NaN`.


```python
ndf = df.copy()
ndf
```

```python
ndf.dropna(thresh=4,      # In a row, it needs at least 4 values is needed, to maintain in df
           inplace = True)
ndf
```

Obviously, rows 3 and 5 has only 2 data, since our `thresh = 4` we need `at least 4 non NaN` data to `maintain` the row.

Defina en qué columnas buscar valores faltantes.

```python
ndf = df.copy()
ndf
```

```python
ndf.dropna(thresh = 5,  #if there is not 5 nan values, the column will be eliminated
           axis = 1,
           inplace = True
           )
ndf
```

#### Solucion 2:

Replace missing values with some other values - Data Imputation



```python
ndf = df.copy()
```

```python
wm = ndf.Weight.mean()
wm
```

```python
ndf['Weight'].fillna(value = wm,
                    inplace = True)
ndf
```

[Datos faltantes: dos grandes problemas con la imputación media](https://www.theanalysisfactor.com/mean-imputation/)

```python
ndf['Length'].fillna(value = ndf.Length.median(),
                    inplace = True)
ndf
```

```python
mm  = ndf.Origin.mode()
mm
```

```python
mm[0]
```

```python
ndf['Origin'].fillna(value = mm[0], #'NoPais',
                    inplace = True)
ndf
```

```python
ndf.isnull().values.any()
```

```python
ndf = df.copy()
ndf
```

Impute particular columns -

```python
ndf.dropna(subset=['Origin', 'Length'],
           inplace = True)
ndf
```

Fill values in each column with favorite operation:

```python
ndf = df.copy()
ndf
```

```python
ndf.Origin.mode()[0]
```

```python
favs = {'Origin': ndf.Origin.mode()[0],
        'Length': ndf['Length'].mean()
       } # setting a dictionary called favs, in which we saves the favorite values.
```

```python
ndf.Origin.fillna(favs['Origin'],
                  inplace=True)
ndf.Length.fillna(favs['Length'],
                  inplace=True)
ndf
```

more on `fillna` method : https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.fillna.html

## Pivot Table


```python
pd.pivot_table(df,
               index='Company',
               values=['Stars','Weight'], # will do the average of each a, b, c, d
               aggfunc={'Stars': np.mean, 'Weight': np.sum},
               margins = True,
               margins_name='|*|Grand Total|*|',
               columns = 'Origin'
              )#.head()
```
